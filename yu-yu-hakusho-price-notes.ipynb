## Hiei, the Stern Completed Listings Snapshot

Source: eBay completed/sold listings query for `hiei the stern yu yu hakusho tcg` (CCG Individual Cards).

Current baseline range in pricing file: $8.00 to $13.00.

```python
hiei_sales = pd.DataFrame([
    {"date": "2026-07-22", "price_usd": 16.47, "title": "Hiei, the Stern - U5 - 1st Edition (Dark Tournament)"},
    {"date": "2026-07-02", "price_usd": 19.99, "title": "Hiei The Stern U5 Dark Tournament Unlimited"},
    {"date": "2026-06-18", "price_usd": 29.99, "title": "Hiei The Stern U5 Dark Tournament Unlimited"},
    {"date": "2026-05-25", "price_usd": 19.99, "title": "Hiei, the Stern - U5 - Unlimited Edition (Dark Tournament)"},
    {"date": "2026-05-23", "price_usd": 23.99, "title": "Hiei The Stern U5 1st Ed Foil Ultra Dark Tournament"},
])

baseline_low = 8.00
baseline_high = 13.00
baseline_mid = (baseline_low + baseline_high) / 2

hiei_sales["delta_vs_mid"] = hiei_sales["price_usd"] - baseline_mid
hiei_sales["status"] = hiei_sales["price_usd"].apply(
    lambda p: "below-range" if p < baseline_low else ("above-range" if p > baseline_high else "in-range")
)

hiei_summary = {
    "sales_count": int(len(hiei_sales)),
    "cumulative_value_usd": float(hiei_sales["price_usd"].sum()),
    "average_price_usd": float(hiei_sales["price_usd"].mean()),
    "median_price_usd": float(hiei_sales["price_usd"].median()),
    "min_price_usd": float(hiei_sales["price_usd"].min()),
    "max_price_usd": float(hiei_sales["price_usd"].max()),
    "baseline_low": baseline_low,
    "baseline_high": baseline_high,
    "baseline_mid": baseline_mid,
    "above_range_count": int((hiei_sales["status"] == "above-range").sum()),
    "in_range_count": int((hiei_sales["status"] == "in-range").sum()),
    "below_range_count": int((hiei_sales["status"] == "below-range").sum())
}

hiei_sales, hiei_summary
```
